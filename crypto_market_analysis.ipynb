# Cryptocurrency Market Cap & Volume Trend Analysis
## Python EDA Project
**Tools:** Python, Pandas, NumPy, Matplotlib, Seaborn

This notebook performs data loading, cleaning, exploratory analysis, trend analysis, ranking, correlation and volatility analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
pd.set_option('display.max_columns', None)
```

```python
path = '../data/raw/crypto_market_data.csv'
df = pd.read_csv(path, parse_dates=['date'])
df.head()
```

```python
print('Shape:', df.shape)
display(df.info())
display(df.describe(include='all'))
```

```python
print('Missing values:')
display(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
```

```python
daily = df.groupby('date', as_index=False).agg(
    total_market_cap=('market_cap','sum'),
    total_volume=('volume_24h','sum')
)
daily.head()
```

```python
plt.figure(figsize=(12,5))
plt.plot(daily['date'], daily['total_market_cap'])
plt.title('Total Crypto Market Capitalization Trend')
plt.xlabel('Date'); plt.ylabel('Market Cap'); plt.xticks(rotation=45)
plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(daily['date'], daily['total_volume'])
plt.title('Total 24H Trading Volume Trend')
plt.xlabel('Date'); plt.ylabel('24H Volume'); plt.xticks(rotation=45)
plt.tight_layout(); plt.show()
```

```python
latest_date = df['date'].max()
top10 = df[df['date']==latest_date].nlargest(10,'market_cap')[['coin_name','symbol','market_cap','volume_24h','market_cap_rank']]
top10
```

```python
corr = df[['close_price','volume_24h','market_cap','daily_return_pct','volatility_pct']].corr()
plt.figure(figsize=(8,6)); sns.heatmap(corr, annot=True, fmt='.2f')
plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

```python
vol = (df.groupby(['symbol','coin_name'], as_index=False)['volatility_pct']
       .mean().sort_values('volatility_pct', ascending=False).head(10))
vol
```

```python
df['month'] = df['date'].dt.to_period('M').astype(str)
monthly = df.groupby('month', as_index=False).agg(
    market_cap=('market_cap','sum'), volume=('volume_24h','sum'),
    avg_return=('daily_return_pct','mean')
)
monthly.tail()
```
